```python
import pandas as pd
import matplotlib.pyplot as plt
from datasets import load_dataset

dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

```python
df_us_jobs = df[df['job_country'] == 'United States'].copy()
```

```python
df_us_jobs['job_posted_month'] = df_us_jobs['job_posted_date'].dt.to_period('M')
```

```python
df_monthly_postings = df_us_jobs.groupby(by='job_posted_month').count()['job_title_short'].reset_index(name='postings_count')
df_monthly_postings
```

```python
df_us_merged = pd.merge(df_us_jobs, df_monthly_postings, how='right', on='job_posted_month')
```

```python
df_us_merged.head()
```

```python
df_not_remote_jobs = df[df['job_work_from_home'] == False].copy()
df_not_remote_salary = df_not_remote_jobs.groupby('job_title_short')['salary_year_avg'].mean().reset_index()
df_not_remote_salary.rename(columns={'salary_year_avg': 'salary_not_remote_avg'}, inplace=True)
df_not_remote_merged = pd.merge(df_not_remote_jobs, df_not_remote_salary, how='right', on='job_title_short')
df_not_remote_merged.head()
```

```python
df_company_salary = df.groupby('company_name')['salary_year_avg'].mean().reset_index()
df_company_salary
```

```python
df_company_remote = df[df['job_work_from_home'] == True].groupby('company_name').size().reset_index(name='remote_count')
df_company_remote
```

```python
df_company_merged = pd.merge(df_company_salary, df_company_remote, how='inner', on='company_name')
df_company_merged
```
