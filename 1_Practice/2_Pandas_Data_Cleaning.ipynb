```python
import pandas as pd
import matplotlib.pyplot as plt
from datasets import load_dataset

dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

```python
df_remove_none = df.copy()

df_remove_none = df_remove_none.dropna(subset='salary_year_avg')
print('Before: ', len(df))
print('After: ', len(df_remove_none))
print('Difference: ', len(df) - len(df_remove_none))
```

```python
df_remove_duplicate = df.copy()
df_remove_duplicate.drop_duplicates(subset='job_location', inplace=True)
print('Before: ', len(df))
print('After: ', len(df_remove_duplicate))
print('Difference: ', len(df) - len(df_remove_duplicate))
```
