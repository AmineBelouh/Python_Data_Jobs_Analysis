```python
import pandas as pd
import matplotlib.pyplot as plt
from datasets import load_dataset

dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

```python
df['job_posted_date_str'] = df['job_posted_date'].apply(lambda posted_date: posted_date.strftime('%Y-%m-%d'))
```

```python
df[['job_posted_date', 'job_posted_date_str']]
type(df['job_posted_date_str'][0])
```

```python
from datetime import datetime

current_date = datetime.now()

df['days_since_posted'] = df['job_posted_date'].apply(lambda date: (current_date - date).days)
```

```python
df[['job_posted_date', 'days_since_posted']].head()
```

```python
df_filtered = df[pd.notna(df['salary_year_avg'])].copy()
```

```python
df_filtered['salary_category'] = df_filtered['salary_year_avg'].apply(lambda salary: 'Low' if salary < 60000 else ('High' if salary > 100000 else 'Medium'))
```
