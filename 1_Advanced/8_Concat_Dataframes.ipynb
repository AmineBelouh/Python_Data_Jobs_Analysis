```python
import pandas as pd
import matplotlib.pyplot as plt
from datasets import load_dataset

dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

```python
df_DA = df[df['job_title_short'] == 'Data Analyst'].copy()
df_DS = df[df['job_title_short'] == 'Data Scientist'].copy()
```

```python
combined_df = pd.concat([df_DA, df_DS], ignore_index=True)
```

```python
combined_df
```

```python
df['job_posted_month'] = df['job_posted_date'].dt.month
```

```python
oct_df = df[df['job_posted_month'] == 10].copy()
nov_df = df[df['job_posted_month'] == 11].copy()
dec_df = df[df['job_posted_month'] == 12].copy()
```

```python
combined_df = pd.concat([oct_df, nov_df, dec_df], ignore_index=True)
combined_df
```
